```python
import pandas as pd
from IPython.display import display

# 1. Configuration de l'affichage Pandas
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.unicode.east_asian_width', True)

# 2. Chargement du dataset (format JSON Lines)
file_path = r'C:\Users\USER\Desktop\pfe\dialogue\bigdata\dataset_final_nlp_v2.jsonl'

try:
    # Lecture avec lines=True
    df = pd.read_json(file_path, lines=True)

    # Nettoyage des noms de colonnes
    df.columns = [col.split('.')[-1] for col in df.columns]

    # 3. Correction globale de la faute de frappe
    # On inclut TOUTES les colonnes susceptibles de contenir le mot
    TEXT_COLUMNS = [
        'instruction', 
        'issue_type', 
        'response',  
        'sentiment_label'
    ]

    for col in TEXT_COLUMNS:
        if col in df.columns:
            # Conversion en string pour éviter les erreurs si la colonne contient des NaN
            df[col] = df[col].astype(str).str.replace('التجوال', 'الجوال', regex=False)
            # On remet les 'None' ou 'nan' en vrais NaN si nécessaire
            df[col] = df[col].replace(['None', 'nan'], pd.NA)
            df[col] = df[col].apply(lambda x: f"USER: عسلامة | {x}" if str(x).startswith("BOT:") else x)
            df['instruction'] = df['instruction'].apply(
            lambda x: x.replace('BOT: إن شاء الله نكونو فدناك', 'USER: يعطيك الصحة | BOT: إن شاء الله نكونو فدناك') 
            if 'BOT: إن شاء الله نكونو فدناك' in str(x) else x
            )

    print("✅ Correction 'التجوال' -> 'الجوال' terminée.")


    # 5. Affichage général
    print("✅ Aperçu du dataset (1000 premières lignes) :")
    display(df.head(1000)) 

    print("-" * 50)
    print(f"Structure : {df.shape[0]} lignes, {df.shape[1]} colonnes.")
    print("-" * 50)
    print(df.info())

except Exception as e:
    print(f"❌ Erreur lors du chargement ou du traitement : {e}")
```

```python
import re

def normalize_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+", "", text)  # liens
    text = re.sub(r"@\w+", "", text)            # mentions
    text = re.sub(r"[^\w\s\u0600-\u06FF]", " ", text)  # ponctuation
    text = re.sub(r"\s+", " ", text).strip()
    return text

df['instruction_clean'] = texts.apply(normalize_text)

display(df[['instruction', 'instruction_clean']].head(10))
```

X.4.1 Apprentissage du type de réclamation (issue_type)

```python
import pandas as pd
import re

def final_perfect_clean(text):
    if pd.isna(text): return text
    
    # 1. Normalisation de base
    text = str(text).lower().strip()
    text = re.sub(r'[أإآ]', 'ا', text)
    text = re.sub(r'ة\b', 'ه', text)
    
    # 2. On définit les catégories "Cibles" (INCLUANT le mot corrigé : الجوال)
    target_categories = [
        "استفسار عن التغطيه", "استفسار عن العروض", "استفسار عن الرصيد",
        "مشكله في الجوال", "تاخير في التركيب", "مشكله في الدفع",
        "اعتراض على الفاتوره", "تبديل شريحه", "عطل في الشبكه",
        "انقطاع الانترنت", "عطب في الجهاز", "تغيير الخدمه",
        "بطء في الانترنت", "مشكله في اشاره الويفي"
    ]
    
    # Création du mapping sans espaces pour une comparaison robuste
    # On enlève aussi les voyelles ou signes si nécessaire, mais l'espace est le plus urgent
    mapping = {c.replace(" ", ""): c for c in target_categories}
    
    # 3. Nettoyage agressif de la clé de recherche (Suppression de TOUS les espaces)
    clean_key = re.sub(r'\s+', '', text) 
    clean_key = clean_key.replace("انترنات", "انترنت")
    clean_key = clean_key.replace("وايفاي", "ويفي")
    clean_key = clean_key.replace("التجوال", "الجوال") # Double sécurité
    
    # 4. Comparaison intelligente
    for key, original in mapping.items():
        # On vérifie si la clé sans espace correspond (ex: "مشكلهفيالجوال")
        if clean_key == key or key in clean_key or clean_key in key:
            return original
            
    # 5. Si aucune correspondance, on renvoie le texte avec des espaces normaux
    return re.sub(r'\s+', ' ', text).strip()

# Application
df['issue_type_clean'] = df['issue_type'].apply(final_perfect_clean)

# Vérification
print(df['issue_type_clean'].value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf,
    y_issue,
    test_size=0.2,
    random_state=42,
    stratify=y_issue
)
```

X.4.2 Apprentissage du sentiment (sentiment_label)

```python
import pandas as pd
import re

def normalize_sentiment_fixed(text):
    if pd.isna(text): return text
    
    # 1. Mise en texte et suppression des espaces internes
    text = str(text).replace(" ", "")
    
    # 2. Normalisation des caractères Arabes
    text = re.sub(r'[أإآ]', 'ا', text)
    text = re.sub(r'ة\b', 'ه', text)
    
    # 3. Recherche par mot-clé (sans espaces)
    if "سلبي" in text:
        return "سلبي"
    elif "محايد" in text:
        return "محايد"
    elif "ايجابي" in text:
        return "إيجابي"
    
    return text

# Application
df['sentiment_label'] = df['sentiment_label'].apply(normalize_sentiment_fixed)

# Vérification finale
print(df['sentiment_label'].value_counts())
```

```python
# On garde uniquement les lignes utiles
df_sentiment = df[['instruction_clean', 'sentiment_label']].dropna()

print("📊 Distribution des sentiments :")
print(df_sentiment['sentiment_label'].value_counts())
print("Shape :", df_sentiment.shape)
```

```python
X_text = df_sentiment['instruction_clean']        # Texte brut
y_sentiment = df_sentiment['sentiment_label']  # Label sentiment
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_sentiment = TfidfVectorizer(
    analyzer='word',
    ngram_range=(1,2),
    max_features=10000
)

X_tfidf_sentiment = tfidf_sentiment.fit_transform(X_text)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf_sentiment,
    y_sentiment,
    test_size=0.2,
    random_state=42,
    stratify=y_sentiment
)
```

```python
from sklearn.linear_model import LogisticRegression

sentiment_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced'
)

sentiment_model.fit(X_train, y_train)
```

```python
import joblib

# Sauvegarde du modèle
joblib.dump(sentiment_model, "sentiment_model.pkl")

# Sauvegarde du vectorizer TF-IDF
joblib.dump(tfidf_sentiment, "tfidf_sentiment.pkl")

print("✅ Modèle et vectorizer sauvegardés")
```

```python
from sklearn.metrics import classification_report, confusion_matrix

y_pred = sentiment_model.predict(X_test)

print("📊 Résultats — Analyse du sentiment")
print(classification_report(y_test, y_pred))

print("📉 Matrice de confusion :")
print(confusion_matrix(y_test, y_pred))
```

X.4.3 Apprentissage du service impacté (service_type)

```python
import pandas as pd
import re

def normalize_service_type_final(text):
    if pd.isna(text): return text
    
    # 1. Mise en minuscules et nettoyage strict
    # On retire tout ce qui n'est pas alphanumérique pour éviter les surprises
    text_raw = str(text).lower()
    clean_key = re.sub(r'[^a-z0-9]', '', text_raw)
    
    # 2. Mapping de correspondance (Ordre stratégique)
    # On traite les cas spécifiques "Réseau/G" en premier
    if "5g" in clean_key or "4g" in clean_key or "reseau" in clean_key:
        return "5G/Réseau"
    
    if "admin" in clean_key:
        return "Administrative"
    if "bill" in clean_key:
        return "Billing"
    if "vdsl" in clean_key:
        return "VDSL"
    if "adsl" in clean_key:
        return "ADSL"
    if "mobile" in clean_key:
        return "Mobile"
    if "fibre" in clean_key:
        return "Fibre"
    if "internet" in clean_key:
        return "Internet"
    if "fixe" in clean_key:
        return "Fixe"
    
    # Retourne le texte original propre si aucune règle ne correspond
    return text.strip().capitalize()

# Application
df['service_type_clean'] = df['service_type'].apply(normalize_service_type_final)

# Vérification
print(df['service_type_clean'].value_counts())
```

```python
# On garde uniquement les colonnes utiles
df_service = df[['instruction_clean', 'service_type_clean']].dropna()

print(df_service.shape)
print(df_service['service_type_clean'].value_counts())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

X_text = df_service['instruction_clean']
y_service = df_service['service_type_clean']

tfidf_service = TfidfVectorizer(
    analyzer='word',
    ngram_range=(1,2),
    max_features=10000
)

X_tfidf_service = tfidf_service.fit_transform(X_text)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf_service,
    y_service,
    test_size=0.2,
    random_state=42,
    stratify=y_service
)
```

```python
from sklearn.linear_model import LogisticRegression

service_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced'
)

service_model.fit(X_train, y_train)
```

```python
import joblib

# Sauvegarde du modèle
joblib.dump(service_model, "service_model.pkl")

# Sauvegarde du vectorizer TF-IDF
joblib.dump(tfidf_service, "tfidf_service.pkl")

print("✅ Modèle et vectorizer sauvegardés")
```

```python
from sklearn.metrics import classification_report

y_pred_service = service_model.predict(X_test)

print("📊 Résultats - Service impacté (service_type)")
print(classification_report(y_test, y_pred_service))
```

```python
def predict_service_smart(text):
    # Prédiction ML
    vect = tfidf_service.transform([text])
    ml_pred = service_model.predict(vect)[0]

    # Règle sémantique prioritaire
    text_clean = text.lower()
    
    if "انترنت" in text_clean or "internet" in text_clean:
        return "Internet"

    return ml_pred
```

X.4.4 Apprentissage de la localisation impactée
(location_wilaya, location_delegation)


```python
import pandas as pd
import re

def final_geo_cleaner(text):
    if pd.isna(text): return text
    
    # 1. Normalisation Arabe de base
    text = str(text).lower().strip()
    text = re.sub(r'[أإآ]', 'ا', text)
    text = re.sub(r'ة\b', 'ه', text)
    
    # 2. Clé de comparaison : on supprime TOUT (espaces, shadda, etc.)
    # Cela transforme "م كثر" en "مكثر" et "القصري ن" en "القصرين"
    clean_key = re.sub(r'\s+', '', text)
    clean_key = re.sub(r'[ّ]', '', clean_key) # Optionnel: enlève les accents

    # 3. Liste de référence (versions sans espaces pour la comparaison)
    reference_names = [
        "تونس", "سوسه", "جندوبه", "صفاقس", "توزر", "بنزرت", "مدنين", "باجه", "نابل", "القيروان",
        "طريقتنيور", "طبرقه", "حمامسوسه", "منزلبورقيبه", "دقاش", "جرجيس", "تستور", 
        "حيالخضراء", "بوحجله", "الحمامات", "ساقيهالداير", "ساقيهالزيت", "سبيطله", 
        "سكره", "رادس", "مكثر", "الشابه", "الفحص", "الرقاب", "قصرهلال", "مارث", 
        "المتلوي", "السرس", "دوز", "غمراسن", "طبربه", "اريانه", "القصرين", "بنعرس", 
        "سليانه", "المهديه", "زغوان", "سيديبوزيد", "المنستير", "قابس", "الكاف", "قفصه", "منوبه", "قبلي", "تطاوين"
    ]
    
    # Dictionnaire pour restaurer les espaces corrects
    official_map = {
        "طريقتنيور": "طريق تنيور",
        "حمامسوسه": "حمّام سوسة",
        "منزلبورقيبه": "منزل بورقيبة",
        "حيالخضراء": "حي الخضراء",
        "ساقيهالداير": "ساقية الداير",
        "ساقيهالزيت": "ساقية الزيت",
        "قصرهلال": "قصر هلال",
        "سيديبوزيد": "سيدي بوزيد",
        "بنعرس": "بن عروس"
    }

    # 4. Matching de sécurité
    if clean_key in reference_names:
        # Si on a une version officielle avec espace, on l'utilise, sinon on rend la clé propre
        return official_map.get(clean_key, clean_key)
            
    # 5. Si aucune correspondance, on renvoie quand même la clé sans espaces parasites
    # (C'est ce qui règle le cas "م كثر" -> "مكثر")
    return clean_key

# Application
df['location_wilaya'] = df['location_wilaya'].apply(final_geo_cleaner)
df['location_delegation'] = df['location_delegation'].apply(final_geo_cleaner)

# Vérification
print(df['location_delegation'].value_counts())
print(df['location_wilaya'].value_counts())
```

```python
# ✅ Correction des incohérences Délégation ↔ Wilaya
# Certaines délégations du dataset sont rattachées à une mauvaise wilaya.
# Ce mapping officiel corrige location_wilaya en se basant sur location_delegation.

DELEGATION_TO_WILAYA = {
    # Tunis
    "طريق تنيور":        "تونس",
    "حي الخضراء":        "تونس",
    # Ariana
    "سكره":              "اريانه",
    # Ben Arous
    "بن عروس":           "بن عروس",
    "رادس":              "بن عروس",
    # Manouba
    "طبربه":             "منوبه",
    # Nabeul
    "الحمامات":          "نابل",
    # Zaghouan
    "الفحص":             "زغوان",
    # Bizerte
    "منزل بورقيبة":      "بنزرت",
    # Béja
    "تستور":             "باجه",
    # Jendouba
    "طبرقه":             "جندوبه",
    # Kef
    "السرس":             "الكاف",
    # Siliana
    "مكثر":              "سليانه",
    # Sousse
    "حمّام سوسة":        "سوسه",
    # Monastir
    "قصر هلال":          "المنستير",
    # Mahdia
    "الشابه":            "المهديه",
    # Sfax
    "ساقية الداير":      "صفاقس",
    "ساقية الزيت":       "صفاقس",
    # Kairouan
    "بوحجله":            "القيروان",
    # Kasserine
    "سبيطله":            "القصرين",
    # Sidi Bouzid
    "الرقاب":            "سيدي بوزيد",
    # Gabès
    "مارث":              "قابس",
    # Medenine
    "جرجيس":             "مدنين",
    # Tataouine
    "غمراسن":            "تطاوين",
    # Gafsa
    "المتلوي":           "قفصه",
    # Tozeur
    "دقاش":              "توزر",
    # Kebili
    "دوز":               "قبلي",
}

def fix_wilaya_from_delegation(row):
    delegation = row['location_delegation']
    wilaya     = row['location_wilaya']
    correct_wilaya = DELEGATION_TO_WILAYA.get(str(delegation))
    if correct_wilaya and correct_wilaya != str(wilaya):
        return correct_wilaya
    return wilaya

# Application
before_counts = df['location_wilaya'].value_counts().to_dict()
df['location_wilaya'] = df.apply(fix_wilaya_from_delegation, axis=1)
after_counts  = df['location_wilaya'].value_counts().to_dict()

# Rapport des corrections
corrections = {
    w: (after_counts.get(w, 0) - before_counts.get(w, 0))
    for w in set(list(before_counts.keys()) + list(after_counts.keys()))
    if after_counts.get(w, 0) != before_counts.get(w, 0)
}
print("✅ Corrections délégation → wilaya appliquées.")
if corrections:
    print("\n📊 Lignes modifiées par wilaya :")
    for w, delta in sorted(corrections.items(), key=lambda x: -abs(x[1])):
        sign = "+" if delta > 0 else ""
        print(f"   {w}: {sign}{delta}")
else:
    print("   Aucun écart détecté (données déjà cohérentes).")
```

```python
# Dataset ML pour la localisation
df_geo = df[['instruction_clean', 'location_wilaya', 'location_delegation']].dropna()

print(df_geo.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_geo = TfidfVectorizer(
    analyzer='word',
    ngram_range=(1,2),
    max_features=8000
)

X_geo = tfidf_geo.fit_transform(df_geo['instruction_clean'])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

y_wilaya = df_geo['location_wilaya']

X_train, X_test, y_train, y_test = train_test_split(
    X_geo,
    y_wilaya,
    test_size=0.2,
    random_state=42,
    stratify=y_wilaya
)

wilaya_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced'
)

wilaya_model.fit(X_train, y_train)

y_pred = wilaya_model.predict(X_test)

print("📊 Résultats — Prédiction de la Wilaya")
print(classification_report(y_test, y_pred))
```

```python
y_delegation = df_geo['location_delegation']

X_train, X_test, y_train, y_test = train_test_split(
    X_geo,
    y_delegation,
    test_size=0.2,
    random_state=42,
    stratify=y_delegation
)

delegation_model = LogisticRegression(
    max_iter=1200,
    class_weight='balanced'
)

delegation_model.fit(X_train, y_train)

y_pred = delegation_model.predict(X_test)

print("📊 Résultats — Prédiction de la Délégation")
print(classification_report(y_test, y_pred))
```

```python
def predict_wilaya(text):
    vect = tfidf_geo.transform([text])
    return wilaya_model.predict(vect)[0]
```

```python
def predict_delegation(text):
    vect = tfidf_geo.transform([text])
    return delegation_model.predict(vect)[0]
```

X.5 Apprentissage des actions et réponses associées

```python
import re
import pandas as pd

# 1. On crée un dictionnaire de référence "propre" (SANS ESPACES) -> "Label correct"
raw_mapping = {
    'إعادة تفعيل الخط': 'إعادة تفعيل الخط',
    'تثبت من حالة الجوال': 'تثبت من حالة الجوال',
    'متابعة حالة الطلب': 'متابعة حالة الطلب',
    'تثبت من تغطية الفيبر': 'تثبت من تغطية الفيبر',
    'تقديم معلومات عن العروض': 'تقديم معلومات عن العروض',
    'تثبت من حالة الشبكة': 'تثبت من حالة الشبكة',
    'اختبار سرعة التدفق': 'اختبار سرعة التدفق',
    'تشخيص تقني': 'تشخيص تقني',
    'تقديم كود USSD': 'تقديم كود USSD',
    'تبديل المودم': 'تبديل المودم',
    'تقديم نصيحة تقنية': 'تقديم نصيحة تقنية',
    'استخراج تفاصيل الفاتورة': 'استخراج تفاصيل الفاتورة',
    'تحويل نوع الخدمة': 'تحويل نوع الخدمة',
    'مد الحريف بموقع فرع': 'مد الحريف بموقع فرع'
}

# On génère un dictionnaire de mapping normalisé (clé sans espaces et sans Alif Hamza)
def normalize_key(t):
    if not isinstance(t, str): return ""
    t = re.sub(r'\s+', '', t) # Supprime tous les espaces
    t = t.replace('أ', 'ا').replace('إ', 'ا').replace('آ', 'ا') # Normalise les Alif
    t = t.replace('التجوال', 'الجوال') # Gestion spécifique de votre cas
    return t

normalized_labels = {normalize_key(k): v for k, v in raw_mapping.items()}

# 2. Fonction de nettoyage intelligente
def clean_label_robust(text):
    if pd.isna(text) or str(text).strip().lower() == 'none':
        return None
    
    # Normalisation de la donnée entrante (on enlève tout espace)
    current_key = normalize_key(str(text))
    
    # On cherche dans notre dictionnaire normalisé
    return normalized_labels.get(current_key, text)

# 3. Application sur les colonnes
cols_to_fix = ['suggested_action', 'issue_type_clean', 'sentiment_label', 'service_type_clean']

for col in cols_to_fix:
    if col in df.columns:
        df[col] = df[col].apply(clean_label_robust)

# Vérification
print(f"✅ Classes finales : {df['suggested_action'].nunique()}")
print(df['suggested_action'].value_counts())
```

```python
# Sélection des colonnes utiles pour la stratégie
df_strategy = df[[
    'issue_type_clean',
    'sentiment_label',
    'service_type_clean',
    'suggested_action',
    'response'
]].dropna().copy()

print(df_strategy.shape)
df_strategy.head()
```

```python
df_strategy['context_key'] = (
    df_strategy['issue_type_clean'] + " | " +
    df_strategy['sentiment_label'] + " | " +
    df_strategy['service_type_clean']
)

df_strategy[['context_key', 'suggested_action', 'response']].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_action = TfidfVectorizer(
    analyzer='word',
    ngram_range=(1,2),
    max_features=3000
)

X_action = tfidf_action.fit_transform(df_strategy['context_key'])
y_action = df_strategy['suggested_action']

```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

X_train, X_test, y_train, y_test = train_test_split(
    X_action, y_action,
    test_size=0.2,
    random_state=42,
    stratify=y_action
)

action_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced'
)

action_model.fit(X_train, y_train)

y_pred_action = action_model.predict(X_test)

print("📊 Résultats – Apprentissage des actions")
print(classification_report(y_test, y_pred_action))
```

```python
response_counts = df_strategy['response'].value_counts()
valid_responses = response_counts[response_counts >= 10].index

df_resp = df_strategy[df_strategy['response'].isin(valid_responses)]

X_resp = tfidf_action.transform(df_resp['context_key'])
y_resp = df_resp['response']

response_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced'
)

response_model.fit(X_resp, y_resp)
```

```python
def predict_action(issue, sentiment, service):
    context = f"ISSUE_{issue} | SENT_{sentiment} | SERV_{service}"
    vect = tfidf_action.transform([context])
    return action_model.predict(vect)[0]

def predict_response(issue, sentiment, service):
    context = f"ISSUE_{issue} | SENT_{sentiment} | SERV_{service}"
    vect = tfidf_action.transform([context])
    return response_model.predict(vect)[0]
```

```python
predict_response(
    issue="عطل في الشبكه",
    sentiment="سلبي",
    service="5G/Réseau"
)
```

X.6 Construction du moteur de décision intelligent

```python
CRITICAL_ISSUES = [
    "عطل في الشبكه",
    "انقطاع الانترنت",
    "مشكله في الدفع",
    "اعتراض على الفاتوره"
]

NEGATIVE_SENTIMENT = "سلبي"
```

```python
def decision_engine(row):
    issue     = row['issue_type_clean']
    sentiment = row['sentiment_label']
    action    = row['suggested_action']

    if action is None or str(action).strip() == "":
        return "escalade_agent_humain"

    if sentiment == NEGATIVE_SENTIMENT and issue in CRITICAL_ISSUES:
        return "escalade_agent_humain"

    return "resolution_automatique"
```

```python
df['decision'] = df.apply(decision_engine, axis=1)
df['decision'].value_counts()
```

X.7 Production des sorties IA et préparation des couches supérieures

1️⃣ Sécurisation des valeurs

```python
def safe_value(val, default="غير محدد"):
    if val is None or str(val).strip().lower() in ["none", "nan", ""]:
        return default
    return val
```

```python
def end_conversation_message():
    return "إن شاء الله نكونو فدناك، وإذا عندك أي استفسار آخر مرحبا بيك في أي وقت."
```

2️⃣ Sortie IA structurée

```python
def generate_ai_output(row):
    return {
        "issue_type": safe_value(row['issue_type_clean']),
        "service_type": safe_value(row['service_type_clean']),
        "sentiment": safe_value(row['sentiment_label']),
        "region": safe_value(row['location_wilaya']),
        "recommended_action": safe_value(row['suggested_action']),
        "decision": safe_value(row['decision'], "resolution_automatique"),
        "end_message": end_conversation_message()
    }
```

```python
def generate_ai_output(text):
    """
    Génère une sortie IA structurée à partir d'une réclamation textuelle
    """

    # Vectorisation du texte
    vect_issue     = tfidf.transform([text])
    vect_sentiment = tfidf_sentiment.transform([text])
    vect_service   = tfidf_service.transform([text])

    # Prédictions ML
    issue_pred     = issue_model.predict(vect_issue)[0]
    sentiment_pred = sentiment_model.predict(vect_sentiment)[0]
    service_pred   = service_model.predict(vect_service)[0]

    # Décision intelligente
    if sentiment_pred == "سلبي" and issue_pred in [
        "عطل في الشبكه", "انقطاع الانترنت", "مشكله في الدفع"
    ]:
        decision = "escalade_agent_humain"
    else:
        decision = "reponse_automatique"

    # Sortie structurée
    output = {
        "issue_type": issue_pred,
        "service_type": service_pred,
        "sentiment": sentiment_pred,
        "decision": decision
    }

    return output
```
